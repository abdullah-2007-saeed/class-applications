## Online order system

```python
class oos:
    def __init__(self,order_id,price,quantity = 0):
        self.order_id = order_id
        self.price = price
        self.quantity = quantity
    def quantity_update(self,num):
        self.quantity += num
    def bill(self):
        total = self.price*self.quantity
        if total >= 3000:
           total = total*0.90
        else:  
            total = total
        print(f'your order({self.order_id}) has total bill of {total}')
    
        
```

```python
c1 = oos(12121,4000)
c2 = oos(13131,2000)
c3 = oos(12345,5000)
```

```python
cartes ={
    1:c1,
    2:c2,
    3:c3
}
```

```python
while True:
    print('select cart')
    cart = int(input('enter cart serial num:'))
    name = cartes[cart]
    
    print('1.update qunatity 2.bill 3.exit')
    num = int(input('enter num:'))
    if num == 1 :
        numb = int(input("enter quantity:"))
        name.quantity_update(numb)
    if num == 2:
        name.bill()
    if num == 3:
        print('Thanks for visiting!!!')
        break
```
