## fuel monitor

```python
class vehicle:
    def __init__(self, vehicle_name, fuel_level):
        self.vehicle_name = vehicle_name
        self.fuel_level = fuel_level

    def add_fuel(self, amount):
        if amount > 0:
            self.fuel_level += amount
            print(f"{amount} units fuel added.")
        else:
            print("Invalid fuel amount.")

    def consume_fuel(self, amount):
        if amount <= self.fuel_level:
            self.fuel_level -= amount
            print(f"{amount} units fuel consumed.")
        else:
            print("Not enough fuel to consume.")

        if self.fuel_level < 10:
            print("Warning: Low fuel level!")

    def display_status(self):
        print(f"Vehicle: {self.vehicle_name}")
        print(f"Fuel Level: {self.fuel_level}")
```

```python
car = vehicle("Toyota Corolla", 15)

car.consume_fuel(6)
car.add_fuel(5)
car.consume_fuel(8)

car.display_status()
```
