## course regestration system

```python
class Course:
    def __init__(self, course_name, max_students):
        self.course_name = course_name
        self.max_students = max_students
        self.enrolled_students = 0

    def enroll_student(self):
        if self.enrolled_students < self.max_students:
            self.enrolled_students += 1
            print("Student enrolled successfully.")
        else:
            print("Enrollment failed: Course is full.")

    def display_status(self):
        print(f"Course: {self.course_name}")
        print(f"Enrolled Students: {self.enrolled_students}/{self.max_students}")
```

```python
course1 = Course("Python Programming", 2)
```

```python
while True:
    print('1.enroll student 2.course status 3.exit')
    num = int(input('enter num:'))
    if num == 1:
        course1.enroll_student()
    if num == 2:
        course1.display_status()
    if num == 3:
        print('Thanks for visiting!')
        break
```
