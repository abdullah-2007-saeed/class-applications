```python
class FeeAccount:
    def __init__(self, student_name, total_fee):
        self.student_name = student_name
        self.total_fee = total_fee
        self.paid_fee = 0

    def pay_fee(self, amount):
        if amount <= 0:
            print("Invalid payment amount.")
            return
        else:
            self.paid_fee += amount
            print(f"{amount} fee paid successfully.")

    def remaining_fee(self):
        return self.total_fee - self.paid_fee

    def display_status(self):
        print(f"Student: {self.student_name}")
        print(f"Total fee: {self.total_fee}")
        print(f"Paid fee: {self.paid_fee}")
        print(f"Remaining fee: {self.remaining_fee()}")
```

```python
s1 = FeeAccount("Ali Khan", 50000)
```

```python
while True:
    print('1.pay fee 2.display fee status 3.exit')
    num = int(input('enter number:'))
    if num == 1:
              f = int(input('enter fee you want to pay:'))
              s1.pay_fee(f)
    elif num == 2:
              s1.display_status()
    elif num == 3:
           print('thanks for visinting')
           break
    else:
        print('invalid input')     
```
