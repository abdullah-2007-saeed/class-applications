## Smart device controller

```python
class sdc:
    def __init__(self,device_name,status = 'off'):
        self.device_name = device_name
        self.status = status
    def device_status(self):
        s = input('enter on/off:')
        if s == 'on':
            self.status = 'On'
        elif s == 'off':
            self.status = 'Off'
        else:
            print('wrong input')
    def display(self):
        print(f'{self.device_name} is {self.status}')
```

```python
d1 = sdc('drill')
d2 = sdc('grinder')
d3 = sdc('motor')
```

```python
machines = {
    1:d1,
    2:d2,
    3:d3
}
```

```python
while True:
    print('......select machine.....')
    machine = int(input('enter machine serial number'))
    name = machines[machine]
    print('1.on/off 2.display status 3.exit')
    num = input('enter num:')
    if num == '1':
        name.device_status()
    elif num == '2':
        name.display()
    elif num == '3':
        print('exit from mode')
        break
    else:
        print('invalid number')
```
