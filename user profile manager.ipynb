## user profile manager:

```python
class profile_manager:
        def __init__(self,name,email,login_count = 0):
            self.name = name
            self.email = email
            self.login_count = login_count
            self.login_count = 0
        def login(self):   
             username = input('Enter username: ')
             email = input('Enter email: ')
             if username == self.name and email == self.email:
                 print('successfully login')
                 self.login_count += 1
             else:
                 print('invalid credentials')
        def display(self):
            print(f'hello {self.name}/{self.email} your login-count is {self.login_count}')
```

```python
u1 = profile_manager('ali','ali1@gmail.com')
u2 = profile_manager('akbar','a1@gmail.com')
u3 = profile_manager('bilal','b1@gmail.com')
```

```python
users = {
    1 : u1,
    2 : u2,
    3 : u3
}
```

```python
while True:
    user = int(input('enter serial number of user:'))
    name = users[user]
    
    print('1. login 2.info 3.exit')
    num = int(input('enter num:'))
    if num == 1:
        name.login()
    elif num == 2:
        name.display()
    else:
        print('Thanks for visiting')
        break
```
