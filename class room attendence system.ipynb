## class room attendence system

```python
class Student:
    def __init__(self, name, roll_number):
        self.name = name
        self.roll_number = roll_number
        self.is_present = False

    def __str__(self):
        if self.is_present == True:
            status = 'present'
        else:
            status = 'absent'
            
        return f"Roll: {self.roll_number} | Name: {self.name} | Status: {status}"

 
classroom = []
print("--- Register 10 Students ---")
for i in range(1, 11):
    name = input(f"Enter name for student {i}: ")
     
    new_student = Student(name, 100 + i) 
    classroom.append(new_student)

 
print("\n--- Mark Daily Attendance (p for Present, a for Absent) ---")
for student in classroom:
    choice = input(f"Is {student.name} (Roll {student.roll_number}) present? (p/a): ").lower()
    if choice == 'p':
        student.is_present = True
    else:
        student.is_present = False

 
print("\n--- Final Attendance Report ---")
for student in classroom:
    print(student)

 
present_count = 0
for student in classroom:
    if student.is_present:
        present_count += 1

print(f"\nTotal Students Present: {present_count}")
print(f"Total Students Absent: {len(classroom) - present_count}")

print('....class present percentage.....')
print(f'class present percentage is:{(present_count/10)*100}')
```
