## banking system

```python
class bank_acc_management:
    def __init__(self,name,acc_no,balance):
        self.name = name
        self.acc_no = acc_no
        self.balance = balance
    def deposit(self,paymet) :
        self.balance += paymet
        print(f'{paymet}rs added to your account your new balance is {self.balance}rs.')
    def withdrawl(self,paymet):
        if paymet <= self.balance:
            self.balance -= paymet
            print(f'{paymet}rs withdrawl from your account, your new balance is {self.balance}rs.')
        else:
            print("you don't have enough money")
    def send_money(self,user_name,money):
        if self.balance >= money:
                           user_name.balance += money
                           self.balance -= money
        else:
            print('low balance!!!')
                          
    def show_balance(self):
        print(f'your balance is {self.balance}rs.')
    
```

```python
u1 = bank_acc_management('ali',2345,5000)
u2 = bank_acc_management('akbar',1234,5000)
u3 = bank_acc_management('bilal',2345,5000)
```

```python
accounts ={
    2345:u1,
    1234:u2,
    3456:u3
}
```

```python
while True:
    acc = int(input('enter your acc no:'))
    if acc not in accounts:
        print("Invalid Account Number!")
        continue
        
    name = accounts[acc]
    
    print('1.show balance 2.withdrawl 3.deposit 4.send money btw acc 5.exit')
    num = input('enter num:')
    if num == '1':
        name.show_balance()
    elif num == '2':
        money = int(input("enter money you wan to withdrawl:\n"))
        name.withdrawl(money)
    elif num == '3':
        money = int(input('enter money you want to deposit:\n'))
        name.deposit(money)
    elif num == '4':
        acc = int(input('enter acc no:'))
        m = int(input('enter money you want to send:'))
        if acc in accounts.keys():
                print(f"My balance before transfer: {name.balance}")
                account_i = accounts[acc]
                name.send_money(account_i,m)
                print(f"My balance after transfer: {name.balance}")    
        else:
            print('no acc found')
    elif num == '5':
        print('Thanks for visiting!!!')
        break
    else:    
        print('Please enter valid num!!!')
```
