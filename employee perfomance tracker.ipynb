## employee perfomance tracker

```python
class ept:
    def __init__(self,name,monthly_target,achieved_target ='in process',bonus ='no_granted'):
        self.name = name
        self.monthly_target = monthly_target
        self.achieved_target = achieved_target
        self.bonus = bonus
    def target_status(self):
        print("if target achieved enter 'y' otherwise 'n'")
        yn = input('enter y/n:').strip().lower()
        if yn == 'y':
            self.achieved_target = 'achieved'
            print('congratulations!!!')
        elif yn == 'n':
            self.achieved_target = 'in process'
            print('no worry, work hard!!!')
        else:
            print('invalid input!!!')
    def bonus_update(self):
        if self.achieved_target == 'achieved':
            self.bonus = 'granted'
        elif self.achieved_target == 'in process' :
            self.bonus = 'no granted'
        else:
            print('system error!!')
    def display(self):
        print(f'employee named:{self.name} has {self.monthly_target} target status is {self.achieved_target} and bonus status is {self.bonus}.')
            
```

```python
e1 =ept('ali','10 files')
```

```python
while True:
    print('1.check employee status 2.update target status 3.exit')
    num = int(input('enter num:'))
    if num == 1:
        e1.bonus_update()
        e1.display()
    elif num == 2:
        e1.target_status()
        e1.bonus_update()
    elif num == 3:
        print('Bye bye!!!')
        break
    else:
        print('invalid input')
```
