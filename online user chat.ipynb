## online user chat

```python
class ouc:
    def __init__(self,name,status = 'offline'):
        self.name = name
        self.status = status
    def change_status(self):
        if self.status == 'offline':
            self.status = 'Online'
        else:
            self.status = 'offline'
    def display(self):
        print(f'{self.name} is now {self.status}')
            
```

```python
u1 = ouc('ali')
```

```python
while True:
    print('1.change status 2.display status 3.exit')
    num = int(input('enter number:'))
    if num == 1:
              u1.change_status()
    elif num == 2:
               u1.display()
    else:
         print('BYE bye!!!')
         break
              
    
```
