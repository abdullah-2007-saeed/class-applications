## shopping cart item

```python
class cart:
    def __init__(self):
        self.product ={
        'laptop':40000,
        'bag':5000,
        'cream':800,
        'cap':1000,
        'jacket':5000
        }
        self.cart ={}
    def pdisplay(self):
        for key,value in self.product.items():
            print(f'rate of {key} is {value}')
    def add_cart(self,namei):
         if namei in self.product:
            if namei in self.cart:
                self.cart[namei] += 1
                print(f"Added another {namei}.")
            else:
                self.cart[namei] = 1
                print(f"Added {namei} to your cart.")
         else:
            print(f"Sorry, '{namei}' is not available in our store.")

    def remove_cart(self,namei):
        if namei in self.cart:
            del self.cart[namei]
            print(f'{namei} removed from your cart')
        else:
            print('item is not in your cart' )
    def cdisplay(self):
          print("\n--- Current Cart ---")
          if not self.cart:
            print("Your cart is empty.")
          else:
            total = 0
            for name, qty in self.cart.items():
                price = self.product[name] 
                subtotal = price * qty
                total += subtotal
                print(f"{name}: {qty} x {price} = {subtotal}")
            if total >= 15000:
                print(f'your total bill after 10 percent discount is {total * 0.90} ')
            else:
                print(f'your tota bill is {total}')
          print('-----------------------------\n')
```

```python
c = cart()
```

```python
c.pdisplay()
while True:
    print('welcome to store....')
    print('1.add cart  2.remove from cart  3.view cart  4.exit')
    num = input('enter number')
    if num == '1':
        namei = input('enter product name:').lower().strip()
        c.add_cart(namei)
    if num == '2':        
            p = input('enter product name:')
            c.remove_cart(p)
    if num == '3':
        c.cdisplay()
    if num == '4':
        print('thanks for shopping')
        break
    
```
