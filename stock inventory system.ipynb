```python
class Product:
    def __init__(self, product_name, stock_quantity):
        self.product_name = product_name
        self.stock_quantity = stock_quantity

    def add_stock(self, quantity):
        self.stock_quantity += quantity

    def sell_product(self, quantity):
        if quantity <= self.stock_quantity:
            self.stock_quantity -= quantity
            print("Sale successful")
        else:
            print("Insufficient stock")

    def display_stock(self):
        print(self.product_name, "Stock:", self.stock_quantity)
```

```python
p1 = Product("Laptop", 10)
```

```python
while True:
    print('1.buy stock 2.sell stock 3.portfolio 4,exit')
    num = int(input('enter number'))
    if num == 1:
        q = int(input('enter quantity:'))
        p1.add_stock(q)
    elif num == 2:
         q = int(input('enter quantity:'))
         p1.sell_product(q)
    elif num == 3:
        p1.display_stock()
    elif num == 4:
        break
    else:
        print('invalid input')
```
