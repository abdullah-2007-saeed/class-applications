## student attendence system

```python
class cas:
    def __init__(self,name,roll_no,is_present = '-'):
        self.name = name
        self.roll_no = roll_no
        self.is_present = is_present
    def attendence_update(self):
        i = input('enter P/A or L for leave:')
        if i == 'p':
            self.is_present = 'Present'
        elif i == 'a':
            self.is_present = 'Absent'
        elif i == 'l':
            self.is_present = 'On Leave'
        else:
            print('invalid input')
    def display(self):
        print(f'{self.name}/{self.roll_no} is {self.is_present}.')    
```

```python
s1 = cas('ali',1)
s2 = cas('ahmad',2)
s3 = cas('abdullah',3)
```

```python
students ={
    1 : s1,
    2: s2,
    3: s3
}
```

```python
while True:
    print('........enter student roll-no to check or update their attendence........')
    student = int(input('enter roll no:'))
    name = students[student]

    print('1.check attendence 2.update 3.exit')
    num = int(input('enter number:'))
    if num == 1:
        name.display()
    if num == 2:
        name.attendence_update()
    if num == 3:
        print('thanks for visiting on portal')
        break
```
