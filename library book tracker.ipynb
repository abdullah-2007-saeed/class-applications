## library book tracker

```python
class lbt:
    def __init__(self,author_name,book_name,is_issued = 'no',availability = 'yes'):
        self.author_name = author_name
        self.book_name = book_name
        self.is_issued = is_issued
        self.availability = availability
    def issue(self):
        if self.is_issued == 'no':
            self.is_issued = 'yes'
            self.availability = 'no'
            print(f'{self.book_name} is issue to you . Thanks!')
        else:
            print('Book is already issued!!')
    def return_book(self):
        if self.is_issued == 'yes':
            self.is_issued = 'no'
            self.availability = 'yes'
            print(f'{self.book_name} is added. Thanks for reading it')
        else:
            print('check book name again we already have this!!')
    def display(self):
        print(f'{self.book_name} availability status is {self.availability}.')
        
        
```

```python
b1 = lbt('george','1984')
b2 = lbt('iqbal','shikwa')
b3 = lbt('dr.israr','seerat un nabi')
```

```python
books ={
    '1984':b1,
    'shikwa':b2,
    'seeratunabi':b3
}
```

```python
while True:
    book = input('enter book name:')
    if book in books.keys():
         name = books[book]
    else:
        print('book is not available')
        break
    
    print('1.book status 2.for issue 3.for return 4.exit portal')
    num = int(input('enter number:'))
    if num == 1:
        name.display()
    if num == 2:
        name.issue()
    if num == 3:
        name.return_book()
    if num == 4:
        print('BYE bye!!!')
        break
    else:
        print('invalid number')
```
